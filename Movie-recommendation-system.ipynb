```python
import ast
```

```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies['genres'].apply(convert)
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
#Extracting top 3 casts like extracting their name
```

```python
def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movies['cast'].apply(convert3)
```

```python
movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python
# Now extracting director from crew
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break    
    return L
```

```python
movies['crew'].apply(fetch_director)
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
movies.head()
```

```python
#stemming applied like convertimg ['Loving','Loved','Love'] --> ['Love','Love','Love']
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tags'].apply(stem)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
ps.stem('Loving')
```

```python
ps.stem('loved')
```

```python
#Cosine distance applied to find the similarity between movies since we are dealing with a high dimentional data.
#Distance inversly proportional to similarity 
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(vectors)
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
similarity[1]
```

```python
#prevent index loosing
sorted(list(enumerate(similarity[0])), reverse = True, key= lambda x:x[1])[1:6]
```

```python
#Now for recommendation index fetching applied
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse = True, key= lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avatar')
```

```python
recommend('Batman Begins')
```

```python
recommend('Interstellar')
```

```python
recommend('Kung Fu Panda')
```

```python
recommend('The Princess and the Frog')
```
